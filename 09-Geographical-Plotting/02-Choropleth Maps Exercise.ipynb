# Choropleth Maps Exercise

## Plotly Imports

```python
import plotly.graph_objs as go 
from plotly.offline import init_notebook_mode,iplot
init_notebook_mode(connected=True)
import pandas as pd
```

** Import pandas and read the csv file: 2014_World_Power_Consumption**

```python
df = pd.read_csv('2014_World_Power_Consumption')
df.head()
```

** Check the head of the DataFrame. **

```python
df.head()
```

** Referencing the lecture notes, create a Choropleth Plot of the Power Consumption for Countries using the data and layout dictionary. **

```python
data = dict(type = 'choropleth',
            locations = df['Country'],
            locationmode = 'country names',
            colorscale= 'Viridis',
            reversescale = True,
            text= df['Country'],
            z=df['Power Consumption KWH'],
            colorbar = {'title':'Power Consumption KWH'})
```

```python
layout = dict(title='2014 Power Consumption',
geo = dict(showframe=False, projection={'type':'mercator'}))
```

```python
choromap = go.Figure(data = [data],layout = layout)
iplot(choromap,validate=False)
```

## USA Choropleth

** Import the 2012_Election_Data csv file using pandas. **

```python
df1 = pd.read_csv('2012_Election_Data')
```

** Check the head of the DataFrame. **

```python
df1.head()
```

** Now create a plot that displays the Voting-Age Population (VAP) per state. If you later want to play around with other columns, make sure you consider their data type. VAP has already been transformed to a float for you. **

```python
data = dict(type = 'choropleth',
            locations = df1['State Abv'],
            locationmode = 'USA-states',
            colorscale= 'Viridis',
            reversescale = True,
            text= df1['State'],
            z=df1['Voting-Age Population (VAP)'],
            colorbar = {'title':'USA State Election voting age population'})
```

```python
layout = dict(title ='2012 Election Data',
geo =dict(scope='usa', showlakes=True, lakecolor='rgb(85,173,240)'))
```

```python
choromap = go.Figure(data = [data],layout = layout)
iplot(choromap,validate=False)
```
