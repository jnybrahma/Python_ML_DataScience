## Boxplots

```python
#df.iplot(kind='box')
fig = px.box(df)
colors = ['red', 'blue', 'green', 'orange']

for i, box in enumerate(fig.data):
    box.marker.color = colors[i]
fig.show()
```

## 3d Surface

```python
df3 = pd.DataFrame({'x':[1,2,3,4,5],'y':[10,20,30,20,10],'z':[5,4,3,2,1]})
# Use scatter_3d for individual points
fig = px.scatter_3d(df3, x='x', y='y', z='z', color='z',color_continuous_scale='rdylbu')
fig.show()
```

```python
df3 = pd.DataFrame({'x':[1,2,3,4,5],'y':[10,20,30,20,10],'z':[5,4,3,2,1]})
# df3.iplot(kind='surface',colorscale='rdylbu')
```

```python
import plotly.graph_objects as go

df3 = pd.DataFrame({'x':[1,2,3,4,5],'y':[10,20,30,20,10],'z':[5,4,3,2,1]})

# Pivot data into a 2D matrix structure where columns are X, index is Y, and cells are Z
# Note: For this small sample, many intersections will be NaN/empty unless properly gridded
grid_df = df3.pivot(index='y', columns='x', values='z')

# Generate the 3D surface plot
fig = go.Figure(data=[go.Surface(
    x=grid_df.columns, 
    y=grid_df.index, 
    z=grid_df.values, 
    colorscale='rdylbu'
)])

fig.show()
```
